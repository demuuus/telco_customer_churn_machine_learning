## **5. Feature Engineering**

### **5.1 Creating preprocessing pipeline**

```python
# Imports for feature engineering

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder,TargetEncoder, StandardScaler, RobustScaler, KBinsDiscretizer
```

```python
for i in df.columns:
    print(i, df[i].unique())
    print()
```

```python
# cat norminal, cat ordinal, numeric, binning, binary

# binary: dependents, paperless billing, churn
# cat nominal (one hot): onlinesecurity, online backup, device protection, tech support
# cat ordinal: contract, internet service
# numeric: tenure, monthly charges

ordinalPipeline = Pipeline(steps=[
    ("encoder", OrdinalEncoder(
        categories=[
            ['Month-to-month', 'One year', 'Two year'], 
            ["No", "DSL", "Fiber optic"]
        ],
        handle_unknown="use_encoded_value", unknown_value=-1
    ))
])

nominalPipeline = Pipeline(steps=[
    ("encoder", OneHotEncoder(drop="first", sparse_output=False, handle_unknown="ignore"))
])

binaryPipeline = Pipeline(steps=[
    ("encoder", OrdinalEncoder(
        categories=[
            ["No", "Yes"]
        ],
        handle_unknown="use_encoded_value", unknown_value=-1
    ))
])

numericalPipeline = Pipeline(steps=[
    ("scaling", StandardScaler())
])
```

### **5.2 Creating feature transformation object**

```python
from sklearn.compose import ColumnTransformer
```

```python
preprocessing = ColumnTransformer(transformers=[
    ("categorical_nominal", nominalPipeline, ["OnlineSecurity", "OnlineBackup", "DeviceProtection", "TechSupport"]),
    ("categorical_ordinal", ordinalPipeline, ["Contract", "InternetService"]),
    ("numerical", numericalPipeline, ["tenure", "MonthlyCharges"]),
    ("binary", binaryPipeline, ["Dependents", "PaperlessBilling", "Churn"])
], remainder="drop", verbose_feature_names_out=False)
```
