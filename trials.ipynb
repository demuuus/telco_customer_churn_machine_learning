# EDA

```python
# Selected features
features = [
    'Dependents',
    'tenure',
    'OnlineSecurity',
    'OnlineBackup',
    'InternetService',
    'DeviceProtection',
    'TechSupport',
    'Contract',
    'PaperlessBilling',
    'MonthlyCharges',
    'Churn'
]

# Count "No internet service" in each column
no_internet_count = (
    df[features]
    .apply(lambda col: col.astype(str).str.strip().eq('No internet service').sum())
    .sort_values(ascending=False)
)

# Display results
print("Count of 'No internet service' per feature (Descending):")
print(no_internet_count)
```

```python
# Filter rows containing "No internet service" in any selected column
filtered_df = df[
    df[features].eq('No internet service').any(axis=1)
]

# Display all matching rows
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

display(filtered_df)
```

## **Creating preprocessing pipeline**

**Dependents, PaperlessBilling, and Churn** is truly binary (yes/no)

**Tenure & MonthlyCharges** is truly numerical

**OnlineSecurity, OnlineBackup, DeviceProtection, TechSupport** aren't truly binaries. They have three conditions: True (yes), False (no), and Inconclusive (No Internet Connection)

**Contract** is truly ordinals. monthtomonth, 1 year, 2 year

```python
# Imports for feature engineering

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder,TargetEncoder, StandardScaler, RobustScaler, KBinsDiscretizer
```

```python
ordinalPipeline = Pipeline(steps=[
    ("encoder", OrdinalEncoder(
        categories=[
            ['Month-to-month', 'One year', 'Two year']
        ],
        handle_unknown="use_encoded_value", unknown_value=-1
    ))
])

binaryPipeline = Pipeline(steps=[
    ("encoder", OrdinalEncoder(
        categories=[
            ["No", "Yes"]
        ],
        handle_unknown="use_encoded_value", unknown_value=-1
    ))
])
```
